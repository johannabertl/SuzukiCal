```python
table_list = [] 
with pdfplumber.open("efteraar_2026.pdf") as pdf:
    for p in pdf.pages:
        for t in p.extract_tables():
            table_list.append(t)
```

```python
# check which table in table_list contains the string "lok."

for i, table in enumerate(table_list):
    for row in table:
        if any("lok." in str(cell) for cell in row):
            print(f"Table {i} contains 'lok.':")
            print(table)
            table_num = i
            break
```

```python
import pandas as pd

plan = pd.DataFrame(table_list[table_num])  

print(plan)

```

```python
# fix the data
# fix the row with NaNs

# print the first row of the plan DataFrame
print(plan.iloc[0])

plan.iloc[0] = plan.iloc[0].ffill() # forward-fill :) 

print(plan.iloc[0])
```

```python
# make column names
colnames = plan.iloc[0] + "-" + plan.iloc[1] 
print(colnames)

```

```python
colnames.iloc[0] = "date"

plan.columns = colnames
plan.drop([0, 1], inplace=True) 
```

```python
# long format

plan_lang = plan.melt(id_vars = ["date"])

# remove non-existing combinations
plan_lang = plan_lang[plan_lang["value"] != ""]
```

```python
# fix data

# fix dates
plan_lang["date"] = plan_lang["date"].str.replace(r"\*+", "", regex=True)
# (or keep the information in a sepparate column? ?)

```

```python
# fix time, room, comment

plan_lang["time"] = plan_lang["variable"].str[4:9] # characters 4-8
plan_lang["room"] = "lok. " + plan_lang["variable"].str[0:3] # first three characters
plan_lang["comment"] = (
    plan_lang["value"].
    str.replace(r"\n", "", regex=True). # remove newlines (makes next line easier)
    str.extract(r"\((.*?)\)", expand=False). # extract text in parentheses. OBS: expand=False returns a Series instead of a DataFrame (otherwise the next line fails)
    str.replace("#", "", regex=False)) # remove "#"

```

```python
# Fix teacher names in comments

teachers = {
    'B': 'Bodil',
    'T': 'Tine',
    'J': 'Jan',
    'S': 'Søren',
    'I': 'Ina'
}

import re
# replace can take a dictionary as input. Create a dictionary with regex patterns as keys and the corresponding teacher names as values. 
# The regex pattern ensures that we only match whole words (i.e., the abbreviation is not part of another word).
pattern_map = {rf'(?<!\w){re.escape(k)}(?!\w)': v for k, v in teachers.items()}

plan_lang["comment"] = plan_lang["comment"].replace(pattern_map, regex=True)
```

```python
# Add group name to comment

plan_lang["group_name"] = plan_lang["value"].str.replace(r"\((.*?)\)", "", regex=True)
```

```python
# Remove irrelevant columns and combine columns

plan_lang["comment"] = plan_lang["comment"] + ", grupper " + plan_lang["group_name"]
plan_lang["datetime"] = pd.to_datetime(plan_lang["date"] + f"/{year} " + plan_lang["time"], format="%d/%m/%Y %H.%M")
plan_lang = plan_lang.drop(columns = ["variable", "value", "date", "time"])

plan_lang.dtypes
```

```python
# save the dataframe to a csv file
plan_lang.to_csv("plan_lang_efteraar_2026.csv", index=False)
```
